```python
from pathlib import Path

repo_path = Path("Fintech-data")

md_files = list(repo_path.rglob("*.md"))
csv_files = list(repo_path.rglob("*.csv"))

print("Markdown files:")
for f in md_files:
    print(f)
print(len(md_files))

print("\nCSV files:")
for f in csv_files:
    print(f)
```

```python
# Department → allowed roles mapping
DEPARTMENT_ROLE_MAP = {
    "Finance": ["Finance", "C-Level"],
    "HR": ["HR", "C-Level"],
    "Marketing": ["Marketing", "C-Level"],
    "Engineering": ["Engineering", "C-Level"],
    "General": ["Employee", "Finance", "HR", "Marketing", "Engineering", "C-Level"]
}
```

```python
from langchain_community.document_loaders import TextLoader
from pathlib import Path

BASE_PATH = Path("Fintech-data")

text_documents = []

for md_file in BASE_PATH.rglob("*.md"):
    department = md_file.parts[1].title()  # normalize case

    loader = TextLoader(str(md_file), encoding="utf-8")
    docs = loader.load()

    for doc in docs:
        doc.metadata["department"] = department
        doc.metadata["source"] = md_file.name
        doc.metadata["allowed_roles"] = DEPARTMENT_ROLE_MAP.get(department, [])

        text_documents.append(doc)
```

```python
from langchain_community.document_loaders import CSVLoader
from pathlib import Path

BASE_PATH = Path("Fintech-data")

csv_documents = []

for csv_file in BASE_PATH.rglob("*.csv"):
    department = csv_file.parts[1].title()  # normalize case

    loader = CSVLoader(file_path=str(csv_file))
    docs = loader.load()

    for doc in docs:
        doc.metadata["department"] = department
        doc.metadata["source"] = csv_file.name
        doc.metadata["allowed_roles"] = DEPARTMENT_ROLE_MAP.get(department, [])

        csv_documents.append(doc)
```

```python
all_documents = text_documents + csv_documents

print(f"Total documents loaded: {len(all_documents)}")
all_documents[1]

```

```python
import langchain
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", " ", ""]
)
chunked_documents = text_splitter.split_documents(all_documents)
print(f"Total chunks created: {len(chunked_documents)}")
chunked_documents[0]


```

```python
import re
def clean_text(text):
    # Remove markdown artifacts
    text = re.sub(r'[#>*_`]', ' ', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', ' ', text)

    # Remove non-ASCII characters
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)

    # Normalize whitespace
    text = re.sub(r'\s+', ' ', text)

    return text.strip()

```

```python
from langchain_core.documents import Document

cleaned_documents = []

for doc in chunked_documents:
    roles = doc.metadata.get("allowed_roles", [])
    if isinstance(roles, list):
        doc.metadata["allowed_roles"] = ",".join(roles)

    cleaned_doc = Document(
        page_content=clean_text(doc.page_content),
        metadata=doc.metadata
    )
    cleaned_documents.append(cleaned_doc)
print("BEFORE:\n", chunked_documents[0].page_content[:300])
print("\nAFTER:\n", cleaned_documents[0].page_content[:300])
```

```python
for doc in all_documents:
    roles = doc.metadata.get("allowed_roles", [])
    if isinstance(roles, list):
        doc.metadata["allowed_roles"] = ",".join(roles)
```

```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# Embedding model
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Create ChromaDB
vector_db = Chroma.from_documents(
    documents=all_documents, # Removed filter_complex_metadata
    embedding=embedding_model,
    persist_directory="./chroma_db"
)

vector_db.persist()

print("ChromaDB created successfully")
```

```python
def strict_secure_search(query, role, k=3):

    # 1️⃣ Validate role
    if role not in DEPARTMENT_ROLE_MAP:
        return [], "Invalid role."

    allowed_departments = DEPARTMENT_ROLE_MAP[role]

    # 2️⃣ Metadata-filtered similarity search
    results = vector_db.similarity_search(
        query=query,
        k=k,
        filter={
            "department": {
                "$in": allowed_departments
            }
        }
    )

    # 3️⃣ Hard security check (guarantees no leakage)
    for doc in results:
        if doc.metadata.get("department") not in allowed_departments:
            raise Exception("SECURITY BREACH: Unauthorized department access!")

    # 4️⃣ Access decision
    if len(results) == 0:
        return [], "Document is not accessible for your role."

    return results, "Access Granted."
```

```python
results, message = strict_secure_search(
    "what is the quarterly revenue?",
    "Finance"
)

print(message)
print("Results count:", len(results))
```
